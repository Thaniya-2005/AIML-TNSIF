**Data Cleaning & Preprocessing**

Task 1: Import Libraries and Upload Dataset

```python
import pandas as pd
import numpy as np
from google.colab import files

# Prompt to upload file
uploaded = files.upload()

# Ensure a file was uploaded before proceeding
if uploaded:
    student_preprocessing_dataset = list(uploaded.keys())[0]
    df = pd.read_csv(student_preprocessing_dataset)
    print("Dataset loaded successfully!")
else:
    print("No file was uploaded. Please re-run the cell and select a file.")
```

Task 2: Display First 5 and Last 5 Rows

```python
print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())
```

Task 3: Find Total Rows and Columns

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

Task 4: Check Data Types

```python
print("Data Types:")
print(df.dtypes)
```

Task 5: Identify Missing Values

```python
print("Missing Values:")
print(df.isnull().sum())
```

Task 6: Fill Missing Values

```python
# Numerical columns
num_cols = df.select_dtypes(include=np.number).columns

df[num_cols] = df[num_cols].fillna(
    df[num_cols].mean()
)

# Categorical columns
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

Task 7: Remove Duplicate Rows

```python
print("Duplicates Before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates After:", df.duplicated().sum())
```

Task 8: Convert Date to DateTime

```python
if "date" in df.columns:
    df["date"] = pd.to_datetime(
        df["date"],
        errors="coerce"
    )

print(df.dtypes)
```

Task 9: Rename Columns to Lowercase

```python
df.columns = df.columns.str.lower()

print(df.columns)
```

Task 10: Convert Float to Integer

```python
float_cols = df.select_dtypes(include="float").columns

for col in float_cols:
    if df[col].notna().all():
        df[col] = df[col].astype(int)

print(df.dtypes)
```

Task 11: Select Multiple Columns

```python
columns = ["age", "marks"]

existing_cols = [col for col in columns if col in df.columns]

print(df[existing_cols])
```

Task 12: Increase Marks by 10%

```python
if "marks" in df.columns:
    df["marks"] = df["marks"] * 1.10

print(df.head())
```

Task 13: Fix Inconsistent Text

```python
text_cols = ["city", "gender"]

for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.capitalize()

print("Cleaned text preview:")
print(df[["gender", "city"]].head())
```

Task 14: Create Pass/Fail Column

```python
if "marks" in df.columns:
    df["result"] = np.where(
        df["marks"] >= 40,
        "Pass",
        "Fail"
    )

print(df.head())
```

Task 15: Filter Students With Marks > 80

```python
if "marks" in df.columns:
    high_marks = df[df["marks"] > 80]

    print("Students with Marks > 80:")
    print(high_marks)
```

Task 16: One-Hot Encoding

```python
categorical_cols = df.select_dtypes(include="object").columns

df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols
)

print(df_encoded.head())
```

Task 17: Sort Based on Marks

```python
if "marks" in df.columns:
    df = df.sort_values(
        by="marks",
        ascending=False
    )

print(df.head())
```

Task 18: Save Cleaned Dataset

```python
df.to_csv(
    "cleaned_dataset.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
