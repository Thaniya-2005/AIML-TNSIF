**Feature Engineering**

```python
# Libraries
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder
```

```python
# Dataset
data = {
    "Age": [22, 28, 35, 42, 50, 24, np.nan, 38, 60, 31],

    "Salary": [
        25000, 35000, 45000, 55000, 65000,
        np.nan, 40000, 50000, 200000, 38000
    ],

    "Gender": [
        "Male", "Female", "Female", "Male", "Male",
        "Female", "Male", None, "Female", "Male"
    ],

    "City": [
        "Hyderabad", "Delhi", "Mumbai", "Hyderabad", "Delhi",
        "Mumbai", None, "Delhi", "Mumbai", "Hyderabad"
    ],

    "Date": [
        "2026-01-10", "2026-02-15", "2026-03-20", "2026-04-25",
        "2026-05-12", "2026-06-18", "2026-07-05", "2026-08-14",
        "2026-09-22", "2026-10-30"
    ],

    "Purchase_Amount": [
        5000, 7000, 8500, 9000, 12000,
        6000, 7500, 10000, 15000, 8000
    ],

    "Experience": [1, 3, 5, 8, 12, 2, 4, 7, 15, 5],

    "Rating": [3.5, 4.0, 4.5, 4.2, 4.8, 3.8, 4.1, 4.4, 4.9, 4.0]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Problem 1: Handling Missing Values
# Numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

df[numerical_columns] = df[numerical_columns].fillna(
    df[numerical_columns].mean()
)

# Categorical columns
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing Values After Cleaning:")
print(df.isnull().sum())

display(df)
```

```python
# Problem 2: Feature Scaling
df["Salary_Before"] = df["Salary"]
df["Age_Before"] = df["Age"]

standard_scaler = StandardScaler()
minmax_scaler = MinMaxScaler()

df["Salary_Scaled"] = standard_scaler.fit_transform(
    df[["Salary"]]
)

df["Age_Scaled"] = minmax_scaler.fit_transform(
    df[["Age"]]
)

print("Before and After Scaling:")

display(
    df[
        [
            "Salary_Before",
            "Salary_Scaled",
            "Age_Before",
            "Age_Scaled"
        ]
    ]
)
```

```python
# Problem 3: Encoding Categorical Data
# Label Encoding Gender
label_encoder = LabelEncoder()

df["Gender_Encoded"] = label_encoder.fit_transform(
    df["Gender"]
)

print("Gender Encoding:")
display(df[["Gender", "Gender_Encoded"]])

```

```python
# One-Hot Encoding City
if "City" in df.columns:
    df = pd.get_dummies(df, columns=["City"], dtype=int)

print("Dataset After One-Hot Encoding City:")
display(df)
```

```python
# Problem 4: Feature Creation
df["Total_Spending"] = (
    df["Salary"] + df["Purchase_Amount"]
)

print("After Creating Total_Spending:")

display(
    df[
        [
            "Salary",
            "Purchase_Amount",
            "Total_Spending"
        ]
    ]
)
```

```python
# Problem 5: Date Feature Extraction
df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.day_name()

print("Extracted Date Features:")

display(
    df[
        [
            "Date",
            "Year",
            "Month",
            "Day",
            "Weekday"
        ]
    ]
)
```

```python
# Problem 6: Age Binning
bins = [0, 25, 45, np.inf]

labels = [
    "Young",
    "Adult",
    "Senior"
]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)

print("Age Groups:")

display(
    df[
        [
            "Age",
            "Age_Group"
        ]
    ]
)
```

```python
# Problem 7: Salary Outlier Detection Using IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

salary_outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

print("Salary Outliers:")

display(salary_outliers)
```

```python
# Problem 8: Feature Interaction
df["Experience_Salary"] = (
    df["Experience"] * df["Salary"]
)

print("Experience and Salary Interaction:")

display(
    df[
        [
            "Experience",
            "Salary",
            "Experience_Salary"
        ]
    ]
)
```

```python
# Problem 9: Feature Selection
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation with target variable
correlation = numeric_df.corr()["Purchase_Amount"].sort_values(ascending=False)

print("Correlation with Purchase Amount:")
print(correlation)

# Get absolute correlations excluding the target itself and removing NaNs (e.g., constant columns like Year)
feature_correlation = correlation.drop("Purchase_Amount").abs().dropna().sort_values()

print("\nFeature Importance Based on Correlation:")
print(feature_correlation)

# Identify the least important feature automatically
least_important = feature_correlation.idxmin()
print(f"\nLeast Important Feature to drop: '{least_important}'")

# Drop the least important feature
df_final = df.drop(columns=[least_important])

print("\nFinal Dataset:")
display(df_final)
```

```python
# Save the Dataset
df_final.to_csv(
    "feature_engineered_dataset.csv",
    index=False
)

print("Dataset saved successfully!")
```
