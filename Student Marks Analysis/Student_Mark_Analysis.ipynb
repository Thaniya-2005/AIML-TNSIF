**Student Mark Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt

students = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
marks = [78, 85, 92, 67, 88, 95, 72, 81, 90, 76]

data = pd.DataFrame({
    'Student': students,
    'Marks': marks
})

print(data)
```

```python
plt.bar(data['Student'], data['Marks'])

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
highest = data.loc[data['Marks'].idxmax()]

print("Highest Scorer:", highest['Student'])
print("Highest Marks:", highest['Marks'])
```

```python
sorted_data = data.sort_values(by='Marks', ascending=False)

print(sorted_data)
```

```python
plt.bar(sorted_data['Student'], sorted_data['Marks'])

plt.title("Student Marks - Descending Order")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
