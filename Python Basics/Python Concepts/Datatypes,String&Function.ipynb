**Datatypes, Strings & Function - Codes**

```python
# Take a list with n elements and append 100
n = int(input("Enter number of elements: "))

numbers = []

for i in range(n):
    value = int(input("Enter element: "))
    numbers.append(value)

numbers.append(100)

print("List:", numbers)
```

```python
# Reverse a List
numbers = [1, 2, 3, 4, 5]

numbers.reverse()

print("Reversed List:", numbers)
```

```python
# Find Minimum & Maximum values
numbers = [2, 7, 9, 1, 3]

print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
```

```python
# Find Sum of Elements in a List
numbers = [10, 20, 30, 40, 50]

total = sum(numbers)

print("Sum:", total)
```

```python
# Remove Duplicates Using Set
numbers = [1, 2, 2, 3, 4, 4, 5]

unique_numbers = set(numbers)

print("After removing duplicates:", unique_numbers)
```

```python
# Check Whether Two Sets Have Common Elements
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

common = set1.intersection(set2)

if common:
    print("Common elements:", common)
else:
    print("No common elements")
```

```python
# Count Unique Characters in a String
text = input("Enter a string: ")

unique = set(text)

print("Unique Characters:", unique)
print("Number of Unique Characters:", len(unique))
```

```python
# Using Functions
def find_min_max(numbers):
    print("Minimum:", min(numbers))
    print("Maximum:", max(numbers))


def find_sum(numbers):
    print("Sum:", sum(numbers))


def reverse_list(numbers):
    return numbers[::-1]


numbers = [10, 5, 20, 15, 30]

find_min_max(numbers)
find_sum(numbers)

print("Reversed:", reverse_list(numbers))
```
