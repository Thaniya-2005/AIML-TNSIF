**Operators**

Arithmetic Operators:
An arithmetic operator is a symbol used to perform basic math tasks like addition (+), subtraction (-), and multiplication (*) on numbers and special Programming Operators.

```python
a = 10
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

Comparison Operators: Comparison operators are symbols used in programming and math to compare two values, such as == (equal to), > (greater than), and < (less than). They always return a boolean value (either true or false) based on whether the condition is met.

```python
a = 10
b = 5

print(a > b)
print(a < b)
print(a == b)
print(a != b)
```

Logical Operators: Logical operators in Python are reserved keywords used to combine or modify conditional statements, evaluating expressions to return a Boolean value (True or False) or one of the operand values. Python provides exactly three logical operators: AND, OR, and NOT

```python
a = 10
b = 5

print(a > 5 and b < 10)
print(a < 5 or b < 10)
print(not a > 5)
```

Assignment Operator: Assignment operators in Python are specialized symbols used to assign, initialize, or update values stored in variables. The most basic assignment operator is the single equal sign (=), which evaluates the expression or value on its right-hand side and saves it directly to the variable named on its left-hand side

```python
a = 10

a += 5
print(a)

a -= 3
print(a)
```
